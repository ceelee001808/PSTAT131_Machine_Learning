### Question 1:
Define supervised and unsupervised learning. What are the difference(s) between them?

_Type your answer here_

Supervised Learning: Supervised learning is basically when you train a model using labeled data (each input comes with a known ouput/target).
    -The model learns a mapping: X->Y
    -The goal is to predict the correct output for new, unseen data
    -Some examples could be predicting home prices or classifying emails as spam/not-spam

Unsupervised Learning: Unsupervised learning uses unlabelled data (meaning there is no predetermined output).
    - The model tries to discover patterns/structures/relationships that can be in the data
    -The goal is to discover hidden features/groupings
    -Some examples could be finding patterns in user behavior 

### Question 2:
Explain the difference between a regression model and a classification model, specifically in the context of machine learning.

_Type your answer here_

Regression Model: A regression model is used when the output (target variable) is continuous (numerical).

### Question 4:
As discussed, statistical models can be used for different purposes. These purposes can generally be classified into the following three categories. Provide a brief description of each.


Descriptive models: Descriptive models are used to summarize or visualize patterns from the main characteristics of a dataset. The goal is not prediction but to highlight trends or relationships. 
    -Example: Using a histogram to show the distribution of heights in a population

Inferential models: Inferential models are used to draw conclusions about a larger population based on a smaller sample of data. This model type is used to test hypotheses and understand relatiobships between different variables (p-values, confidence intervals, testing)
    -Example: Confucting a t-test to determine if a new medication is significantly more effective than a placebo

Predictive models: Predictive models focus on forecasting future outcomes or unoberserved values. These models emphasis on minimizing prediction error.
    -Example: Using a Random Forest Classifier to estimate the probability that a prospective athlete will succeed at the professional level based on their historical performance metrics.

### Question 5:
Predictive models are frequently used in machine learning, and they can usually be described as either mechanistic or empirically-driven. Answer the following questions.

Define mechanistic. Define empirically-driven. How do these model types differ? How are they similar? 

Mechanistic (parametric) models: assume a specific functional form (linear relationship) because you believe that is how the underlying system actually works. These models are built on pre-existing thoery.

Empirically-Driven Models: these models don't start with theory. Instead, empirically data-driven modles use algorithms to "discover" patterns and function forms directly from the data. They are also flexible and can adpat to complex non-linear shapes.

These models are similare because they both recquire data for validation and aim to minimize the difference between predicted/actual values. Also both use optimization techniques to find the best parameters.

In general, is a mechanistic or empirically-driven model easier to understand? Explain your choice.

In my opinion I believe mechanistic models are significantly easier to understand. This is because they rely on specific mathematical equations. For example in linear regression the slope wil tell you exactly how much the output changes for every unit of input x. Empirically driven models may involve thousands of hidden interactions that make it super difficult to understand.

Describe how the bias-variance tradeoff is related to the use of mechanistic or empirically-driven models.

Mechanistic Models → Higher bias, lower variance. 
    - Might miss true relationships if the real world doesn't follow theory.
    - Low variance because model won't change drastically if you train on different set of data 

Empirically-driven → Lower bias, higher variance.
    - Low bias because they are flexible to fit almost every data shape 
    -High variance (overfitting). Can memorize the noise in your specific data causing the model to perform poorly when applied to unseen data.
